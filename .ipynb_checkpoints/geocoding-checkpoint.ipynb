```python
import pandas as pd
import re
from pathlib import Path
from rapidfuzz import fuzz, process
from geopy.geocoders import Nominatim
from geopy.distance import geodesic
from collections import defaultdict
import time
from tqdm import tqdm
import math
```

```python
# Load data
DATA_DIR = Path("data")
RAW_DIR = DATA_DIR / "raw"
OUTPUT_DIR = Path("output")
GTFS_DIR = RAW_DIR / "ttc_gtfs"

unique_df = pd.read_csv(OUTPUT_DIR / "unique_locations_with_ids.csv")
ml_df = pd.read_csv(OUTPUT_DIR / "ml_dataset_ready.csv")
stops_df = pd.read_csv(GTFS_DIR / "stops.txt")
routes_df = pd.read_csv(GTFS_DIR / "routes.txt")
trips_df = pd.read_csv(GTFS_DIR / "trips.txt")
stop_times_df = pd.read_csv(GTFS_DIR / "stop_times.txt")

# normalization function
COMMON_FIXES = {
    "TRITION": "TRITON",
    "BOROUGHQ": "BOROUGH",
    "BOROUG": "BOROUGH",
    "BOUROUGH": "BOROUGH",
    "KNIFGHTSB": "KNIGHTSBRIDGE",
    "KNIGHTSBR": "KNIGHTSBRIDGE",
    "EDLIOU": "EDILOU",
    "HRONER": "HORNER",
    "MCNICOL": "MCNICOLL",
    "NAIRIN": "NAIRN",
    "ABOT": "ABBOT",
    "ABBEY": "ABBEY",
}

def normalize_location_combined(text):
    if not isinstance(text, str):
        return ""
    text = text.upper().strip()
    for old, new in COMMON_FIXES.items():
        text = text.replace(old, new)

    text = re.sub(r"\bAND\b", "&", text)
    text = re.sub(r"\bAT\b", "&", text)
    text = re.sub(r"\s+", " ", text)
    return text.strip()

def canonical_intersection(text):
    text = normalize_location_combined(text)
    if "&" not in text:
        return None
    parts = text.split("&")
    if len(parts) != 2:
        return None
    streets = []

    for p in parts:
        p = re.sub(
            r"\b(ST|STREET|RD|ROAD|AVE|AV|AVENUE|DR|DRIVE|BLVD|HWY)\b",
            "", p
        )
        p = p.strip()
        if p:
            streets.append(p)

    if len(streets) == 2:
        return "|".join(sorted(streets))
    return None
    
# Drop stops with missing coordinates
stops_df = stops_df.dropna(subset=['stop_lat', 'stop_lon'])
print(f"Loaded {len(stops_df)} GTFS stops with valid coordinates")
print(f"Loaded {len(unique_df)} unique locations")
print(f"Loaded {len(ml_df)} ML rows")

# Extract routes per location from ML dataset (only the 50 routes)
routes_per_location = ml_df.groupby('location_id')['route'].apply(
    lambda x: list(set(x.dropna().astype(str)))).to_dict()

unique_df['routes_list'] = unique_df['location_id'].map(routes_per_location)
unique_df['routes_list'] = unique_df['routes_list'].apply(lambda x: x if isinstance(x, list) else [])

valid_route_ids = set(ml_df['route'].dropna().astype(str).unique())
print(f"Number of routes in bus data: {len(valid_route_ids)}")

# Normalise GTFS stop names and build exact lookup
print("Normalising GTFS stops")
tqdm.pandas(desc="Normalising GTFS")
stops_df['normalized'] = stops_df['stop_name'].progress_apply(normalize_location_combined)

gtfs_lookup = {}  # normalized_key -> (stop_name, lat, lon)
for _, row in stops_df.iterrows():
    key = row['normalized']
    if key and key not in gtfs_lookup:
        gtfs_lookup[key] = (row['stop_name'], row['stop_lat'], row['stop_lon'])

canonical_gtfs_lookup = defaultdict(list)
for _, row in stops_df.iterrows():
    c = canonical_intersection(row["normalized"])
    if c:
        canonical_gtfs_lookup[c].append(
            (
                row["stop_name"],
                row["stop_lat"],
                row["stop_lon"]
            )
        )
        
gtfs_keys = list(gtfs_lookup.keys())
print(f"Unique GTFS normalized keys: {len(gtfs_keys)}")

# Blocking dictionary
blocked = defaultdict(list)
for key in gtfs_keys:
    block = key[:2] if len(key) >= 2 else key
    blocked[block].append(key)

# Build route->stops mapping for ONLY the routes present in bus data
trip_route = dict(zip(trips_df['trip_id'], trips_df['route_id']))

stop_info = {}
for _, row in stops_df.iterrows():
    stop_info[row['stop_id']] = (row['stop_name'], row['normalized'])

route_stops = defaultdict(set)
for _, row in stop_times_df.iterrows():
    trip_id = row['trip_id']
    stop_id = row['stop_id']
    route_id = trip_route.get(trip_id)
    if route_id and str(route_id) in valid_route_ids and stop_id in stop_info:
        _, norm_stop = stop_info[stop_id]
        route_stops[str(route_id)].add(norm_stop)

# Convert route_stops sets to lists for faster iteration
route_stops = {k: list(v) for k, v in route_stops.items()}
print(f"Built route->stops mapping for {len(route_stops)} routes (filtered to {len(valid_route_ids)} bus routes)")

# Helper functions
GEOCODE_FALLBACK = True   

def exact_match(norm_key):
    return gtfs_lookup.get(norm_key, None)
# Cache for route-restricted fuzzy matches
fuzzy_cache = {}

def fuzzy_match(norm_key, route_ids=None, score_cutoff=75):
    cache_key = (norm_key, tuple(sorted(route_ids)) if route_ids else None)
    if cache_key in fuzzy_cache:
        return fuzzy_cache[cache_key]
    
    block = norm_key[:2] if len(norm_key) >= 2 else norm_key
    if route_ids and len(route_ids) > 0:
        candidates = []
        for rid in route_ids:
            candidates.extend(route_stops.get(str(rid), []))
        candidates = list(dict.fromkeys(candidates))
        if not candidates:
            candidates = blocked.get(block, gtfs_keys)
    else:
        candidates = blocked.get(block, gtfs_keys)
    
    if not candidates:
        fuzzy_cache[cache_key] = (None, None, None, None, 0)
        return None, None, None, None, 0
    
    result = process.extractOne(norm_key, candidates, scorer=fuzz.WRatio, score_cutoff=score_cutoff)
    if result:
        best_key, score, _ = result
        stop_name, lat, lon = gtfs_lookup[best_key]
        if lat is not None and lon is not None and math.isfinite(lat) and math.isfinite(lon):
            fuzzy_cache[cache_key] = (stop_name, lat, lon, best_key, score)
            return stop_name, lat, lon, best_key, score
    fuzzy_cache[cache_key] = (None, None, None, None, 0)
    return None, None, None, None, 0

# Cache for full fuzzy
full_fuzzy_cache = {}

def full_fuzzy_match(norm_key, score_cutoff=70):
    cache_key = (norm_key, score_cutoff)
    if cache_key in full_fuzzy_cache:
        return full_fuzzy_cache[cache_key]
    block = norm_key[:2] if len(norm_key) >= 2 else norm_key
    candidates = blocked.get(block, gtfs_keys)
    if not candidates:
        full_fuzzy_cache[cache_key] = (None, None, None, None, 0)
        return None, None, None, None, 0
    result = process.extractOne(norm_key, candidates, scorer=fuzz.WRatio, score_cutoff=score_cutoff)
    if result:
        best_key, score, _ = result
        stop_name, lat, lon = gtfs_lookup[best_key]
        if lat is not None and lon is not None and math.isfinite(lat) and math.isfinite(lon):
            full_fuzzy_cache[cache_key] = (stop_name, lat, lon, best_key, score)
            return stop_name, lat, lon, best_key, score
    full_fuzzy_cache[cache_key] = (None, None, None, None, 0)
    return None, None, None, None, 0

def is_address_like(text):
    return bool(re.search(r'\d', text)) and len(text.split()) >= 2

geolocator = Nominatim(user_agent="ttc_location_updater")
geocode_cache = {}

def geocode_address(address):
    if address in geocode_cache:
        return geocode_cache[address]
    try:
        time.sleep(1)
        loc = geolocator.geocode(f"{address}, Toronto, ON", timeout=10)
        if loc is not None:
            lat, lon = loc.latitude, loc.longitude
            if math.isfinite(lat) and math.isfinite(lon):
                geocode_cache[address] = (lat, lon)
                return lat, lon
    except Exception:
        pass
    geocode_cache[address] = (None, None)
    return None, None

def find_nearest_stop(lat, lon):
    if lat is None or lon is None or not (math.isfinite(lat) and math.isfinite(lon)):
        return None, None, None, None
    min_dist = float('inf')
    best = None
    for _, row in stops_df.iterrows():
        s_lat, s_lon = row['stop_lat'], row['stop_lon']
        if not (math.isfinite(s_lat) and math.isfinite(s_lon)):
            continue
        d = geodesic((lat, lon), (s_lat, s_lon)).meters
        if d < min_dist:
            min_dist = d
            best = row
    if best is not None:
        return best['stop_name'], best['stop_lat'], best['stop_lon'], min_dist
    return None, None, None, None

# Process each unique location
results = []
unmatched_ids = []
geocode_attempts = 0

for idx, row in tqdm(unique_df.iterrows(), total=len(unique_df), desc="Matching locations"):
    loc_id = row['location_id']
    norm_key = row['location']
    raw = row['location']
    routes = row['routes_list']
    best_norm = norm_key
    best_stop_name = None
    best_lat = None
    best_lon = None
    match_type = "unmatched"

    # ---- Tier 1 ----
    match = exact_match(norm_key)
    if match:
        stop_name, lat, lon = match
        best_norm = stop_name
        best_stop_name = stop_name
        best_lat = lat
        best_lon = lon
        match_type = "exact"
        results.append({'location_id': loc_id, 'normalized_best': best_norm, 'gtfs_stop_name': best_stop_name,
            'stop_lat': best_lat, 'stop_lon': best_lon, 'match_type': match_type})
        continue

    # ---- Tier 1.5 Canonical intersection ----
    canonical = canonical_intersection(norm_key)
    if canonical and canonical in canonical_gtfs_lookup:
    
        stop_name, lat, lon = canonical_gtfs_lookup[canonical][0]
    
        best_norm = stop_name
        best_stop_name = stop_name
        best_lat = lat
        best_lon = lon
        match_type = "intersection_exact"
        results.append({
            'location_id': loc_id,
            'normalized_best': best_norm,
            'gtfs_stop_name': best_stop_name,
            'stop_lat': best_lat,
            'stop_lon': best_lon,
            'match_type': match_type
        })
    
        continue
    

    # ---- Tier 2: Route-restricted fuzzy (cutoff 80, then 75) ----
    stop_name, lat, lon, matched_key, score = fuzzy_match(norm_key, route_ids=routes, score_cutoff=80)
    if not stop_name:
        stop_name, lat, lon, matched_key, score = fuzzy_match(norm_key, route_ids=routes, score_cutoff=75)
    if stop_name:
        best_norm = stop_name
        best_stop_name = stop_name
        best_lat = lat
        best_lon = lon
        match_type = f"fuzzy_{score}"
        results.append({'location_id': loc_id, 'normalized_best': best_norm, 'gtfs_stop_name': best_stop_name,
            'stop_lat': best_lat, 'stop_lon': best_lon, 'match_type': match_type})
        continue

    # ---- Tier 3: Full fuzzy (no route restriction, cutoffs 70,65,60) ----
    stop_name, lat, lon, matched_key, score = full_fuzzy_match(norm_key, score_cutoff=70)
    # if not stop_name:
    #     stop_name, lat, lon, matched_key, score = full_fuzzy_match(norm_key, score_cutoff=65)
    # if not stop_name:
    #     stop_name, lat, lon, matched_key, score = full_fuzzy_match(norm_key, score_cutoff=60)
    if stop_name:
        best_norm = stop_name
        best_stop_name = stop_name
        best_lat = lat
        best_lon = lon
        match_type = f"fullfuzzy_{score}"
        results.append({'location_id': loc_id, 'normalized_best': best_norm, 'gtfs_stop_name': best_stop_name,
            'stop_lat': best_lat, 'stop_lon': best_lon, 'match_type': match_type})
        continue

    # ---- Tier 4: Geocoding ----
    if GEOCODE_FALLBACK and (is_address_like(raw) or is_address_like(norm_key)):
        addr = raw if is_address_like(raw) else norm_key
        lat, lon = geocode_address(addr)
        geocode_attempts += 1
        if lat is not None and lon is not None and math.isfinite(lat) and math.isfinite(lon):
            stop_name, s_lat, s_lon, dist = find_nearest_stop(lat, lon)
            if stop_name:
                best_norm = stop_name
                best_stop_name = stop_name
                best_lat = s_lat
                best_lon = s_lon
                match_type = f"geo_{dist:.0f}m"
                results.append({'location_id': loc_id, 'normalized_best': best_norm, 'gtfs_stop_name': best_stop_name,
                    'stop_lat': best_lat, 'stop_lon': best_lon, 'match_type': match_type})
                continue

    # ---- Tier 5: Unmatched ----
    unmatched_ids.append(loc_id)
    results.append({'location_id': loc_id, 'normalized_best': best_norm, 'gtfs_stop_name': best_stop_name,
        'stop_lat': best_lat, 'stop_lon': best_lon, 'match_type': match_type})

print(f"Geocoding attempted for {geocode_attempts} locations (if enabled)")

# Merge and save
update_df = pd.DataFrame(results)
updated_unique = unique_df.merge(update_df, on='location_id', how='left')
updated_unique['location'] = updated_unique['normalized_best']
updated_unique['stop_lat'] = updated_unique['stop_lat']
updated_unique['stop_lon'] = updated_unique['stop_lon']
updated_unique['match_type'] = updated_unique['match_type']
updated_unique.drop(columns=['normalized_best', 'routes_list'], inplace=True, errors='ignore')

updated_unique.to_csv(OUTPUT_DIR / "unique_locations_lookup1.csv", index=False)

print(f"\nUpdated unique locations saved to {OUTPUT_DIR / 'unique_locations_lookup1.csv'}")
print(f"Total locations: {len(updated_unique)}")
print(f"Matched: {len(update_df[update_df['match_type'] != 'unmatched'])}")
print(f"Unmatched: {len(update_df[update_df['match_type'] == 'unmatched'])}")
```

```python
# SECOND PASS MATCHING ONLY UNMATCHED LOCATIONS
# Street Pair Exact -> Fuzzy WRatio -> Save lookup2
DATA_DIR = Path("data")
RAW_DIR = DATA_DIR / "raw"
OUTPUT_DIR = Path("output")
GTFS_DIR = RAW_DIR / "ttc_gtfs"

unique_df = pd.read_csv(OUTPUT_DIR / "unique_locations_lookup1.csv")
stops_df = pd.read_csv(GTFS_DIR / "stops.txt")

stops_df = stops_df.dropna(subset=["stop_lat", "stop_lon"])
print("Total locations:", len(unique_df))

# Get only unmatched
unmatched_df = unique_df[unique_df["match_type"]=="unmatched"].copy()
print("Unmatched before second pass:", len(unmatched_df))

# Normalization
COMMON_FIXES = {
    "TRITION":"TRITON",
    "BOROUGHQ":"BOROUGH",
    "BOROUG":"BOROUGH",
    "BOUROUGH":"BOROUGH",
    "KNIFGHTSB":"KNIGHTSBRIDGE",
    "KNIGHTSBR":"KNIGHTSBRIDGE",
    "EDLIOU":"EDILOU",
    "HRONER":"HORNER",
    "MCNICOL":"MCNICOLL",
    "NAIRIN":"NAIRN",
    "ABOT":"ABBOT"
}

def normalize_location_combined(text):
    if not isinstance(text,str):
        return ""

    text=text.upper().strip()
    for old,new in COMMON_FIXES.items():
        text=text.replace(old,new)
        
    text=re.sub(r"\bAND\b","&", text)
    text=re.sub(r"\bAT\b","&", text)
    text=re.sub(r"\s+"," ",text)
    return text.strip()

# Street pair extraction
def street_pair(text):
    text=normalize_location_combined(text)
    if "&" not in text:
        return None

    parts=text.split("&")
    if len(parts)!=2:
        return None
        
    streets=[]
    for p in parts:
        p=re.sub(
            r"\b(ST|STREET|RD|ROAD|AVE|AV|AVENUE|DR|DRIVE|BLVD|BOULEVARD|HWY)\b",
            "",
            p
        )
        p=p.strip()
        
        if p:
            streets.append(p)

    if len(streets)==2:

        return "|".join(
            sorted(streets)
        )
    return None

# Normalize GTFS
print("Normalizing GTFS")
tqdm.pandas(desc="GTFS normalization")

stops_df["normalized"] = (stops_df["stop_name"].progress_apply(normalize_location_combined))

# Build GTFS lookup
gtfs_lookup={}
for _,row in stops_df.iterrows():
    key=row["normalized"]
    if key not in gtfs_lookup:
        gtfs_lookup[key]=(
            row["stop_name"],
            row["stop_lat"],
            row["stop_lon"]
        )

gtfs_keys=list(gtfs_lookup.keys())
print("GTFS keys:",len(gtfs_keys))

# Build street pair lookup
print("Building street pair lookup")

street_lookup={}
for _,row in stops_df.iterrows():
    key=street_pair(row["stop_name"])

    if key:
        street_lookup[key]=(
            row["stop_name"],
            row["stop_lat"],
            row["stop_lon"]
        )
        
print("Street pair keys:", len(street_lookup))

# Process unmatched
results=[]
for _,row in tqdm(
    unmatched_df.iterrows(),
    total=len(unmatched_df),
    desc="Second pass matching"
):

    loc_id=row["location_id"]
    original=row["location"]
    result={

        "location_id":loc_id,
        "normalized_best":original,
        "gtfs_stop_name":None,
        "stop_lat":None,
        "stop_lon":None,
        "match_type":"unmatched"

    }

    # Tier 1: Street pair exact
    pair=street_pair(original)
    if pair in street_lookup:
        stop,lat,lon = street_lookup[pair]
        result.update({
            "normalized_best":stop,
            "gtfs_stop_name":stop,
            "stop_lat":lat,
            "stop_lon":lon,
            "match_type":"street_pair_exact"
        })

        results.append(result)
        continue

    # Tier 2: Fuzzy WRatio
    norm=normalize_location_combined(original)
    fuzzy_result=process.extractOne(
        norm,
        gtfs_keys,
        scorer=fuzz.WRatio
    )

    if fuzzy_result:
        best_key,score,_=fuzzy_result
        # threshold
        if score >= 75:
            stop,lat,lon = gtfs_lookup[best_key]
            result.update({
                "normalized_best":stop,
                "gtfs_stop_name":stop,
                "stop_lat":lat,
                "stop_lon":lon,
                "match_type":
                    f"second_fuzzy_{score}"
            })

    results.append(result)

# Create second pass dataframe
second_df=pd.DataFrame(results)
print(
    "\nSecond pass matched:",
    len(
        second_df[
            second_df["match_type"]!="unmatched"
        ]
    )
)

# Merge back
updated = unique_df.merge(
    second_df,
    on="location_id",
    how="left",
    suffixes=("","_new")
)

for col in [
    "normalized_best",
    "gtfs_stop_name",
    "stop_lat",
    "stop_lon",
    "match_type"
]:
    
    new_col=col+"_new"
    if new_col in updated.columns:
        updated[col]=(
            updated[new_col]
            .combine_first(
                updated[col]
            )
        )

        updated.drop(
            columns=[new_col],
            inplace=True
        )

updated["location"]=(
    updated["normalized_best"]
    .combine_first(
        updated["location"]
    )
)

updated.drop(
    columns=[
        "normalized_best"
    ],
    inplace=True,
    errors="ignore"
)

# Save
output_file = (OUTPUT_DIR /"unique_locations_lookup2.csv")
updated.to_csv(output_file, index=False)
print("\nSaved:")
print(output_file)
print("\nFinal results:")
print(updated["match_type"].value_counts())
```
